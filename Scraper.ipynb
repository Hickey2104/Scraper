```python
#Modules
import requests
from bs4 import BeautifulSoup
import re
```

```python
#Define scraping and cleaning functions
def scrape_text(url):
    """Fetch and clean visible text from a webpage."""
    print(f"Scraping: {url}")
    try:
        response = requests.get(url, timeout=10)
        response.encoding = 'utf-8'
        soup = BeautifulSoup(response.text, "html.parser")
        text = soup.get_text(separator=" ")
        text = re.sub(r'\s+', ' ', text)
        text = re.sub(r'[^a-zA-Z0-9.,!?\'" ]+', '', text)
        return text.strip()
    except Exception as e:
        print(f"⚠️ Error scraping {url}: {e}")
        return ""

def clean_text(text):
    """Final text cleaning."""
    text = text.lower()
    text = re.sub(r'\s+', ' ', text)
    return text.strip()
```

```python
#Websites that data is scraped from
urls = [
    "https://baseballrulesacademy.com/official-rule/",
    "https://www.mlb.com/official-information/official-rules",
    "https://www.umpirebible.com/OBR16/home.htm",
    "https://www.baseball-reference.com/bullpen/Baseball_rules",
    "https://baseballscouter.com/basic-baseball-rules-for-beginners/",
    "https://www.dugoutdebate.com/baseball-rules-overview/",
    "https://youthbaseballedge.com/basic-baseball-rules/",
    "https://www.rulesofsport.com/sports/baseball.html",
    "https://www.nfhs.org/activities-sports/baseball/",
    "https://ondeck.baseballontario.com/page/3013/official-baseball-rules/15868/official-baseball-rules-900",
    "https://content.mlb.com/documents/2/2/4/305750224/2019_Official_Baseball_Rules_FINAL_.pdf",
    "https://static.wbsc.org/uploads/federations/0/cms/documents/a56bcb85-f358-6057-ab9d-848e17bb46e9.pdf",
    "https://ipmall.law.unh.edu/sites/default/files/hosted_resources/SportsEntLaw_Institute/League%20rules%2C%20regulations/MajorLeagueRules-2008.pdf",
    "https://www.mlb.com/glossary/rules",
    "https://www.mlb.com/official-information",
    "https://www.mlb.com/glossary/rules/regulation-game",
    "https://www.lib.umd.edu/collections/university-archives/digital/baseball-rules",
    "https://library.si.edu/digital-library/book/spaldingquotsba1886",
    "https://onlinebooks.library.upenn.edu/webbin/book/browse?c=x&key=Baseball%2C+Periodicals&type=lcsubc",
    "https://www.mlb.com/official-information/official-rules/glossary",
    "https://www.mlb.com/official-information/official-rules/foreword",
    "https://www.scorebooklive.com/blog/official-baseball-rules-explained",
    "https://www.umpirebible.com/index.php/rules-of-baseball"
]
```

```python
#Scrap and combine text
all_text = ""
for url in urls:
    page_text = scrape_text(url)
    all_text += "\n\n" + page_text
```

```python
#Cleaning
cleaned_data = clean_text(all_text)
```

```python
#Save cleaned data
with open("cleaned_baseball_rules.txt", "w", encoding="utf-8") as f:
    f.write(cleaned_data)

print("✅ Done! Combined and cleaned text saved as 'cleaned_baseball_rules.txt'")
```
